## Test Case 6: Function definition and call

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [n for n in range(2, 50) if is_prime(n)]
print(primes)
```

## Test Case 7: Class definition and usage

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def annual(self):
        return self.salary * 12

emps = [Employee("Asha", 55000), Employee("Ravi", 62000)]
for e in emps:
    print(e.name, "->", e.annual())
```

## Test Case 8: Exception handling

```python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError as e:
        return f"Error: {e}"

print(divide(10, 2))
print(divide(10, 0))
```

## Test Case 9: Recursion (Fibonacci)

```python
def fib(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib(n - 1, memo) + fib(n - 2, memo)
    return memo[n]

print([fib(i) for i in range(15)])
```

## Test Case 14: SQL Setup - create in-memory SQLite database with tables

```python
import sqlite3
conn = sqlite3.connect(":memory:")
cur = conn.cursor()

cur.execute('''
CREATE TABLE departments (
    dept_id INTEGER PRIMARY KEY,
    dept_name TEXT
)
''')

cur.execute('''
CREATE TABLE employees (
    emp_id INTEGER PRIMARY KEY,
    emp_name TEXT,
    dept_id INTEGER,
    salary REAL,
    hire_date TEXT,
    FOREIGN KEY (dept_id) REFERENCES departments(dept_id)
)
''')

conn.commit()
print("Tables created: departments, employees")
```

## Test Case 17: SQL - JOIN across tables

```python
join_result = pd.read_sql_query('''
    SELECT e.emp_name, e.salary, d.dept_name
    FROM employees e
    JOIN departments d ON e.dept_id = d.dept_id
    ORDER BY d.dept_name, e.salary DESC
''', conn)
join_result
```

## Test Case 18: SQL - GROUP BY aggregation

```python
agg_result = pd.read_sql_query('''
    SELECT d.dept_name,
           COUNT(e.emp_id) AS headcount,
           ROUND(AVG(e.salary), 2) AS avg_salary,
           SUM(e.salary) AS total_salary
    FROM departments d
    LEFT JOIN employees e ON d.dept_id = e.dept_id
    GROUP BY d.dept_name
    ORDER BY avg_salary DESC
''', conn)
agg_result
```

## Test Case 19: SQL - Subquery

```python
subquery_result = pd.read_sql_query('''
    SELECT emp_name, salary
    FROM employees
    WHERE salary > (SELECT AVG(salary) FROM employees)
''', conn)
subquery_result
```
