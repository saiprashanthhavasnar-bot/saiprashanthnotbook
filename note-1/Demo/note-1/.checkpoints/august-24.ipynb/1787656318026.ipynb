## 5. Graph and Chart Test Cases

```python
# TC-16: Line Graph
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]

plt.figure(figsize=(8, 5))
plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# TC-17: Bar Chart
departments = df["department"].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(departments.index, departments.values)
plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.show()
```

```python
# TC-18: Pie Chart
department_count = df["department"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(department_count.values, labels=department_count.index, autopct="%1.1f%%")
plt.title("Department Distribution")
plt.show()
```

```python
# TC-19: Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(df["employee_id"], df["salary"])
plt.title("Employee ID vs Salary")
plt.xlabel("Employee ID")
plt.ylabel("Salary")
plt.show()
```

```python
# TC-20: Histogram
plt.figure(figsize=(8, 5))
plt.hist(df["salary"], bins=5)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

## 9. Large Data and Performance Tests

```python
# TC-32: Generate large DataFrame

import pandas as pd

large_df = pd.DataFrame({
    "id": range(1, 100001),
    "value": range(1, 100001)
})

print("Rows:", len(large_df))

large_df.head()
```

```python
# TC-33: Performance test
import time

start_time = time.time()

large_df.groupby("value").count()

end_time = time.time()

print("Execution time:", round(end_time - start_time, 4), "seconds")
```

```python
# TC-34: Memory information
memory_mb = large_df.memory_usage(deep=True).sum() / (1024 * 1024)
print(f"DataFrame memory usage: {memory_mb:.2f} MB")
```
